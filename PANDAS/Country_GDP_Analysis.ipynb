```python
import matplotlib.pyplot as plt # visualization
import seaborn as sns # statistics visualization

#plot the graph in the line
%matplotlib inline 
plt.rcParams['figure.figsize'] = 6,2 # rcParams is a parameter comes from plt library where the figure size width = 6 and height = 2

import warnings
warnings.filterwarnings('ignore') # whenever os will update then ignore the os error by using worning message
```

```python
df.columns
```

```python
df['InternetUsers']
```

```python
vis1 = sns.distplot(df['InternetUsers'])
```

```python
vis1 = sns.distplot(df["InternetUsers"]) # univariant analysis - plot the graph using 1 variable is called univariate analysis.
plt.show(vis1)
```

```python
sns.__version__
```

```python
vis2 = sns.displot(df["InternetUsers"]) # univariant analysis - plot the graph using 1 variable is called univariate analysis.
plt.show(vis2)
```

```python
vis3 = sns.distplot(df["InternetUsers"], bins=15)
plt.show(vis3)
```

```python
plt.rcParams['figure.figsize'] = 10,5
vis4 = sns.boxplot(data = df, x = 'IncomeGroup', y = 'BirthRate') # bivariate - plot the graph with two variables is called bivarate
plt.show(vis4)
```

```python
# Outlier = in statics, the outlier is the data point which is very far from other observations
# outlier is also called as ANOMALY DETECTION
# min, 25%, mean/median, 75%
```

```python
vis5 = sns.lmplot(data = df, x = 'InternetUsers', y = 'BirthRate') 
plt.show(vis5)
```

```python
vis6 = sns.lmplot(data = df, x = 'InternetUsers', y = 'BirthRate', fit_reg=False) 
plt.show(vis6)
```

```python
vis7 = sns.lmplot(data = df, x = 'InternetUsers', y = 'BirthRate', fit_reg=True) 
plt.show(vis7)
```

```python
vis8 = sns.lmplot(data = df, x = 'InternetUsers', y = 'BirthRate', fit_reg=True, hue='IncomeGroup') 
plt.show(vis8)
```

```python
vis9 = sns.lmplot(data = df, x = 'InternetUsers', y = 'BirthRate', fit_reg=False, hue='IncomeGroup') 
plt.show(vis9)
```
