# Function with arguments

```python
def add(x, y): # defining the function add() with formal arguments x and y
    c = x + y
    print(c)

add(5, 6) # calling function add() with 5, 6
```

```python
# Function with return type
def add(x, y): 
    c = x + y
    return c

add(5, 6)
```

```python
# Function with return type
def add(x, y, z): 
    c = x + y + z
    return c

add(5, 6, 7)
```

```python
# Function with return type
def add(x, y, z, n): 
    c = x + y + z + n
    return c

add(5, 6, 7, 8)
```

```python
def greet(): 
    print('Hello.')
    print('Good morning.')

def add(x, y): 
    c = x + y
    return c


def sub(x, y):
    d = x - y
    return d

greet()
add(5, 6)
sub(5, 6)
```

```python
def add_sub(x, y):
    c = x + y
    d = x - y
    return c, d

add_sub(5, 6)
```

```python
add_sub(4, 5)
```

```python
type(add_sub)
```

```python
result = add_sub(4, 5)
print(result)
print(type(result))
```

```python
result, result1 = add_sub(4, 5)
print(result)
print(result1)
print(type(result))
print(type(result1))
```

```python
def add_sub_mul(x, y):
    c = x + y
    d = x - y
    e = x * y
    return c, d, e

add, sub, mul = add_sub_mul(5, 6)
print('addition of 2 numbers = ', add)
print('substraction of 2 numbers = ', sub)
print('multiplication of 2 nunbers = ', mul)
```

```python
def add_sub_mul(x, y):
    c = x + y
    d = x - y
    e = x * y
    return c, d, e

add, sub, mul = add_sub_mul(5, 6)

add
sub
mul
```
